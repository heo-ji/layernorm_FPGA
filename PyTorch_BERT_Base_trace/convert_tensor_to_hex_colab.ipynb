# Tensor to Hex 변환

Google Drive의 `MyDrive/bert_output`에 있는 `tensor_2_hex.py`로 `qnli`, `qqp`, `rte` tensor를 변환합니다.

```python
from google.colab import drive

drive.mount('/content/drive')
%cd /content/drive/MyDrive/bert_output
```

```python
import os

assert os.path.isfile('tensor_2_hex.py'), 'bert_output 안에 tensor_2_hex.py가 없습니다.'
assert os.path.isdir('GLUEtask_tensor'), 'bert_output 안에 GLUEtask_tensor 폴더가 없습니다.'

for task in ('qnli', 'qqp', 'rte'):
    tensor_dir = os.path.join('GLUEtask_tensor', task)
    assert os.path.isdir(tensor_dir), f'{tensor_dir} 폴더가 없습니다.'
    os.makedirs(task, exist_ok=True)
    pt_count = sum(
        name.endswith('.pt')
        for root, _, files in os.walk(tensor_dir)
        for name in files
    )
    print(f'{task}: {pt_count}개 .pt 파일 ({tensor_dir})')
```

```python
# 생성된 결과 폴더 확인
import os

for task in ('qnli', 'qqp', 'rte'):
    output_dirs = [
        name for name in os.listdir(task)
        if name in ('hex_txt_32x768_k1', 'hex_txt_32x768_k2', 'hex_txt_32x768_k5')
    ]
    print(f'{task}: {sorted(output_dirs)}')
```
